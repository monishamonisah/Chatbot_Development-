# Week 9 – Chatbot Development using Ollama

**Domain:** Education  
**Model:** `llama3.2`

## Objective
Build an interactive command-line chatbot, maintain context within a session, and save timestamped conversations.

## Architecture
User question → Python chatbot → Ollama API → Local LLM (`llama3.2`) → Response → Terminal + CSV logs

Run the command below from the project root folder.

```python
!python ../chatbot.py
```

## 4. Test plan
Try at least one question from each type:
- Factual: What is an algorithm?
- Reasoning: I have two exams tomorrow. How can I plan my study time?
- Coding: Write a Python program to check whether a number is even or odd.
- General conversation: Hello! Can you help me study?
- Context: First say `My subject is Python.` Then ask `Which subject did I mention?`

Check `../dataset/chat_history.csv`, `../dataset/responses.csv`, and `../dataset/chatbot_logs.txt` after testing.

## 5. Observations
Complete this section after running the chatbot.

**Strengths**
- Local model integration through Ollama.
- Follow-up context during one running session.
- Timestamped CSV and text logs.

**Limitations**
- The model can generate inaccurate answers.
- Response speed depends on hardware.
- In-memory conversation history resets after restart.

**Improvements**
- Review incorrect answers and refine the system prompt.
- Add input validation and more evaluation questions.
- Integrate LangChain in the next week.
